# OpenNeuro EEG dataset downloader

This notebook downloads OpenNeuro dataset `ds004504` into the repository's `dataset` directory. The dataset contains the EEG recordings consumed by the preprocessing and causal-graph pipeline.

The download is resumable: rerunning the cell checks existing files and retrieves missing content. A network connection is required, and the complete dataset may take substantial time and disk space. Change `dataset_id` to select another OpenNeuro dataset or `target_dir` to store it elsewhere.

```python
from pathlib import Path
import openneuro

dataset_id = "ds004504"
target_dir = Path("dataset")
target_dir.mkdir(parents=True, exist_ok=True)

# Create the destination before starting the resumable download.
print(f"Downloading dataset {dataset_id}...")
openneuro.download(
    dataset=dataset_id,
    target_dir=target_dir,
)
print(f"Dataset downloaded to: {target_dir.resolve()}")
```
