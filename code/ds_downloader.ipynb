```python
from pathlib import Path
import openneuro

dataset_id = "ds004504"
target_dir = Path("dataset")
target_dir.mkdir(parents=True, exist_ok=True)

print(f"Download del dataset {dataset_id}...")
openneuro.download(
    dataset=dataset_id,
    target_dir=target_dir,
)
print(f"Dataset scaricato in: {target_dir.resolve()}")
```
